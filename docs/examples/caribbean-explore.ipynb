## STEP 2: Run the case

The cell below solves each of the case's 3 decks from 2026 to 2050 and draws their well-to-wake emissions, their freight rate and the fuel they burn, side by side.

```python
# @title Run the case and see the results
import sys
from pathlib import Path

import matplotlib.pyplot as plt

# The workshop's figure and the functions that fill it, shared with the
# Examples pages: docs/workshop/_results.py.
sys.path.insert(0, str(Path("docs/workshop").resolve()))
from _results import compare, plot_cases, print_summary, run_case, scale, summarise
from _solver import guard_interior_point

guard_interior_point()

NAVS = {
    "no policy": Path("simulations/examples/example_5/example_5_no_policy.nav"),
    "current rules": Path("simulations/examples/example_5/example_5_current_rules.nav"),
    "extended": Path("simulations/examples/example_5/example_5_extended.nav"),
}
cases = {}
for label, nav in NAVS.items():
    print(f"--- {label}")
    cases[label] = summarise(run_case(nav))
    print_summary(cases[label])
plot_cases(cases)
plt.show()

# The levers below move this deck:
NAV, ref = NAVS["extended"], cases["extended"]
```

## STEP 3: Move one lever

Each number multiplies what the deck already says, so 1.0 is the reference. Set one, run the cell, and compare with the extended deck. As an example, the fuel oil price is doubled.

```python
# ===========================================================================
#  YOUR SCENARIO. Every number is a MULTIPLIER: 1.0 is the deck as it is.
# ===========================================================================
OIL_PRICE                  = 2.0   # fuel oil price
GFS_TIER_1_PRICE           = 1.0   # price of the global fuel standard's first tier
AMERICAS_ELECTRICITY_PRICE = 1.0   # electricity price in the Americas, where e-fuels are made

LEVERS = {
          "fossil_fuel_oil_price_global": OIL_PRICE,
          "gfs_tier_1_price": GFS_TIER_1_PRICE,
          "electricity_cost_americas_mid": AMERICAS_ELECTRICITY_PRICE}
TWEAKS = [scale(f, v) for f, v in LEVERS.items() if v != 1.0]


def your_scenario(m):
    for t in TWEAKS:
        t(m)


compare(NAV, ref, your_scenario)
plt.show()
```
