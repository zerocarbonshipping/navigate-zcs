## STEP 2: Run the case

The cell below solves each of the case's 2 decks from 2026 to 2050 and draws their well-to-wake emissions, their freight rate and the fuel they burn, side by side.

```python
# @title Run the case and see the results
import sys
from pathlib import Path

import matplotlib.pyplot as plt

# The workshop's figure and the functions that fill it, shared with the
# Examples pages: docs/workshop/_results.py.
sys.path.insert(0, str(Path("docs/workshop").resolve()))
from _results import compare, plot_cases, print_summary, run_case, scale, summarise
from _solver import guard_interior_point

guard_interior_point()

NAVS = {
    "reference": Path("simulations/examples/example_6/example_6_reference.nav"),
    "CO2 price": Path("simulations/examples/example_6/example_6_co2_price.nav"),
}
cases = {}
for label, nav in NAVS.items():
    print(f"--- {label}")
    cases[label] = summarise(run_case(nav))
    print_summary(cases[label])
plot_cases(cases)
plt.show()

# The levers below move this deck:
NAV, ref = NAVS["CO2 price"], cases["CO2 price"]
```

## STEP 3: Move one lever

Each number multiplies what the deck already says, so 1.0 is the reference. Set one, run the cell, and compare with the CO2 price deck. As an example, the fuel oil price is doubled.

```python
# ===========================================================================
#  YOUR SCENARIO. Every number is a MULTIPLIER: 1.0 is the deck as it is.
# ===========================================================================
OIL_PRICE  = 2.0   # fuel oil price
CO2_PRICE  = 1.0   # Germany's CO2 price, 65 to 150 USD/t (INVENTED)
HULL_PRICE = 1.0   # newbuild price of the hull, 6 M USD (INVENTED)

LEVERS = {
          "fossil_fuel_oil_price_global": OIL_PRICE,
          "germany_co2_price": CO2_PRICE,
          "rhine_hull_capex": HULL_PRICE}
TWEAKS = [scale(f, v) for f, v in LEVERS.items() if v != 1.0]


def your_scenario(m):
    for t in TWEAKS:
        t(m)


compare(NAV, ref, your_scenario)
plt.show()
```
